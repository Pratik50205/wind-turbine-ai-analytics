# Hill of Towie Wind Farm — Exploratory Data Analysis & Empirical Power Curves

This notebook demonstrates empirical power curve derivation, Betz limit validation, and turbine utilization analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

PROJECT_ROOT = Path('..').resolve()
df_perf = pd.read_csv(PROJECT_ROOT / 'outputs' / '04_performance_analysis' / 'turbine_performance.csv')
df_pc = pd.read_csv(PROJECT_ROOT / 'outputs' / '04_performance_analysis' / 'power_curve_data.csv')
df_perf.head()
```

```python
# Plot empirical power curve vs reference rated capacity
plt.figure(figsize=(10, 6))
ws_col = 'wind_speed_midpoint' if 'wind_speed_midpoint' in df_pc.columns else 'wind_speed_bin'
pw_col = 'curve_power_kw' if 'curve_power_kw' in df_pc.columns else 'power_mean'
plt.plot(df_pc[ws_col], df_pc[pw_col], '-o', color='#0284c7', lw=2.5, label='Measured Power Curve')
plt.axhline(y=2300, color='#f59e0b', lw=2, linestyle='--', label='Siemens SWT-2.3-VS-82 (2300 kW)')
plt.title('Hill of Towie Farm Empirical Power Curve (IEC 61400-12)', fontsize=12, fontweight='bold')
plt.xlabel('Wind Speed (m/s)')
plt.ylabel('Active Power (kW)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```
